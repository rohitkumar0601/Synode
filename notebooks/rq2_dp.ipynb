# RQ2: Differential Privacy Trade-Offs (Epsilon Sweep & Pareto Fronts)

**Research Question 2 (RQ2):** *What is the empirical trade-off between differential privacy ($\varepsilon$) and data utility/fidelity?*

This notebook analyzes:
1. MST $\varepsilon \in \{0.1, 1.0, 10.0\}$ sweep curves across fidelity (`quality_score`), utility (`tstr_mean`), and empirical privacy risks.
2. Privacy budget composition and public vs. private bounds fallback.
3. Pareto front shift: how adding DP generators expands or alters the utility-fidelity vs. privacy Pareto frontier.
4. Nominal DP audit (PrivBayes) vs. mathematically verified $\varepsilon$-DP.
